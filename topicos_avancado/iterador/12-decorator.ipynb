Decorator
- Modificam ou estendem o comportamento de funções existentes de forma reutilizável.
- Adicionam funcionalidades a funções existentes sem modificar seu código diretamente.

```python
def moldura(funcao):
    def nova_versao():
        print("==========")
        funcao()
        print("==========")
    return nova_versao


@moldura
def bom_dia():
    print("Bom dia!")


@moldura
def boa_noite():
    print("Boa noite!")

boa_noite()
bom_dia()
```
